```python
import cvxpy as cp
import numpy as np
```

```python
# Problem data.
m = 30
n = 20
np.random.seed(1)
A = np.random.randn(m, n)
b = np.random.randn(m)

# Construct the problem.
x = cp.Variable(n)
objective = cp.Minimize(cp.sum_squares(A @ x - b))
constraints = [0 <= x, x <= 1]
prob = cp.Problem(objective, constraints)

# The optimal objective value is returned by `prob.solve()`.
result = prob.solve()
# The optimal value for x is stored in `x.value`.
print(x.value)
# The optimal Lagrange multiplier for a constraint is stored in
# `constraint.dual_value`.
print(constraints[0].dual_value)
```

```python
# Problem 5
def g(A):
    return cp.sum(cp.norm(A, axis=0))


x = cp.Variable(5)
prob = cp.Problem(cp.Minimize(
    g(A - cp.outer(x, np.ones(10)))
))
res = prob.solve()
print(res)



```

```python
# Problem 6
def h(A):
    return cp.sigma_max(A)


x = cp.Variable(5)
prob = cp.Problem(cp.Minimize(
    h(A - cp.outer(x, np.ones(10)))
))
res = prob.solve()
print(res)


```
