```python
import re
import json
```

```python
def extract_from_script(script_text):
    # Regex to extract the 'series' array
    match = re.search(r"series:\s*(\[\{.*\}\])", script_text, re.DOTALL)
    if match:
        series_data = match.group(1)

        # Preprocess the data: Convert JS object syntax to JSON
        # Replace single quotes with double quotes
        series_data = re.sub(r"'", '"', series_data)
        
        # Add double quotes around unquoted keys
        series_data = re.sub(r"(\w+):", r'"\1":', series_data)
        
        # Convert the series data to a Python list
        # print(series_data)
        parsed_data = json.loads(series_data)
        # print(parsed_data)
        
    
        # Example: Accessing individual series
        for series in parsed_data:
            print(f"Company: {series['name']}, Data: {series['data']}")
    else:
    
        
        print("Series data not found!")
    return parsed_data
```
